## 7. Data Cleaning - Orders Dataset

In this section, we start cleaning the main orders dataset.

We will handle duplicate records, missing values, incorrect data types,
inconsistent text formatting, and invalid numerical values.

We will create a separate working copy to preserve the original dataset.

```python
orders_clean = orders.copy()

print("Original Shape:", orders.shape)
print("Working Copy Shape:", orders_clean.shape)
```

### 7.2 Remove Exact Duplicates

We remove rows that are completely identical across all columns.

This prevents duplicated records from affecting the analysis.

```python
print("Duplicates Before Cleaning:", orders_clean.duplicated().sum())

orders_clean = orders_clean.drop_duplicates()

print("Duplicates After Cleaning:", orders_clean.duplicated().sum())
print("New Shape:", orders_clean.shape)
```

### 7.3 Standardize Text Values

We standardize categorical text columns by removing extra spaces
and applying consistent capitalization.

This helps avoid treating the same category as different values.

```python
text_columns = ['Order_Status', 'Payment_Method']

for col in text_columns:
    orders_clean[col] = orders_clean[col].str.strip().str.title()

print("Order Status Values:")
print(orders_clean['Order_Status'].value_counts())

print("\nPayment Method Values:")
print(orders_clean['Payment_Method'].value_counts())
```

### 7.4 Convert Data Types

We convert date columns into datetime format and numerical columns
into appropriate numeric data types.

Invalid values that cannot be converted will be treated as missing values.

```python
# Convert date columns
orders_clean['Order_Date'] = pd.to_datetime(
    orders_clean['Order_Date'], errors='coerce'
)

orders_clean['Ship_Date'] = pd.to_datetime(
    orders_clean['Ship_Date'], errors='coerce'
)

# Convert Unit_Price to numeric
orders_clean['Unit_Price'] = pd.to_numeric(
    orders_clean['Unit_Price'], errors='coerce'
)

print(orders_clean.dtypes)
```

### 7.5 Handle Invalid Numerical Values

We inspect numerical columns and identify values that are outside
the expected business rules.

Negative quantities and discounts outside the range of 0 to 1
are considered invalid and will be replaced with missing values.

```python
# Handle negative quantities
orders_clean.loc[
    orders_clean['Quantity'] < 0, 'Quantity'
] = np.nan

# Handle invalid discounts
orders_clean.loc[
    (orders_clean['Discount'] < 0) |
    (orders_clean['Discount'] > 1),
    'Discount'
] = np.nan

# Convert Quantity to nullable integer
orders_clean['Quantity'] = orders_clean['Quantity'].astype('Int64')

print("Invalid Quantities:", (orders_clean['Quantity'] < 0).sum())
print("Invalid Discounts:", (
    (orders_clean['Discount'] < 0) |
    (orders_clean['Discount'] > 1)
).sum())
```

### 7.6 Validate Shipping Dates

We check whether shipping dates occur before order dates.

Invalid shipping dates are converted to missing values.
Missing shipping dates for pending and cancelled orders are preserved,
as these orders may not have been shipped.

```python
invalid_shipping = (
    orders_clean['Ship_Date'].notna() &
    orders_clean['Order_Date'].notna() &
    (orders_clean['Ship_Date'] < orders_clean['Order_Date'])
)

print("Invalid Shipping Dates:", invalid_shipping.sum())

orders_clean.loc[invalid_shipping, 'Ship_Date'] = pd.NaT
```

### 7.7 Final Check

After cleaning, we review the dataset structure, missing values,
and duplicate records to verify the cleaning operations.

```python
print("Cleaned Orders Shape:", orders_clean.shape)

print("\nMissing Values:")
print(orders_clean.isnull().sum())

print("\nExact Duplicates:", orders_clean.duplicated().sum())

print("\nData Types:")
print(orders_clean.dtypes)

display(orders_clean.head())
```

## 8. Data Cleaning - New Daily Batch

In this section, we clean the new daily batch dataset.

This dataset contains inconsistent numerical formats, missing values,
duplicate records, and inconsistent text formatting.

We will prepare the dataset for integration with the main orders dataset.

```python
new_batch_clean = new_batch.copy()

print("Original Shape:", new_batch.shape)
print("Working Copy Shape:", new_batch_clean.shape)
```

### 8.2 Remove Exact Duplicates

We remove completely duplicated rows to avoid repeated records.

```python
print("Duplicates Before:", new_batch_clean.duplicated().sum())

new_batch_clean = new_batch_clean.drop_duplicates()

print("Duplicates After:", new_batch_clean.duplicated().sum())

print("New Shape:", new_batch_clean.shape)
```

### 8.3 Clean Quantity Column

The Quantity column contains numerical values stored as text,
including values with the "pcs" suffix.

We remove the unnecessary text, convert the column to numeric,
and handle negative quantities as invalid values.

```python
new_batch_clean['Quantity'] = (
    new_batch_clean['Quantity']
    .astype(str)
    .str.replace(r'\s*pcs\s*$', '', regex=True)
)

new_batch_clean['Quantity'] = pd.to_numeric(
    new_batch_clean['Quantity'], errors='coerce'
)

new_batch_clean.loc[
    new_batch_clean['Quantity'] < 0, 'Quantity'
] = np.nan

new_batch_clean['Quantity'] = new_batch_clean['Quantity'].astype('Int64')

print(new_batch_clean['Quantity'].value_counts(dropna=False))
print("\nData Type:", new_batch_clean['Quantity'].dtype)
```

### 8.4 Clean Discount Column

The Discount column contains values represented in different formats,
such as decimal numbers and percentage strings.

We convert all values into a consistent decimal format.
Invalid discounts outside the range of 0 to 1 are treated as missing.

```python
discount_text = new_batch_clean['Discount'].astype(str).str.strip()

is_percentage = discount_text.str.contains('%', na=False)

discount_values = pd.to_numeric(
    discount_text.str.replace('%', '', regex=False),
    errors='coerce'
)

discount_values.loc[is_percentage] = (
    discount_values.loc[is_percentage] / 100
)

new_batch_clean['Discount'] = discount_values

new_batch_clean.loc[
    (new_batch_clean['Discount'] < 0) |
    (new_batch_clean['Discount'] > 1),
    'Discount'
] = np.nan

print(new_batch_clean['Discount'].value_counts(dropna=False))
```

### 8.5 Standardize Text Columns

We remove unnecessary spaces and standardize capitalization
in categorical columns.

This ensures consistent category names across the dataset.

```python
text_columns = ['Order_Status', 'Payment_Method']

for col in text_columns:
    new_batch_clean[col] = (
        new_batch_clean[col]
        .str.strip()
        .str.title()
    )

print("Order Status:")
print(new_batch_clean['Order_Status'].value_counts())

print("\nPayment Method:")
print(new_batch_clean['Payment_Method'].value_counts())
```

### 8.6 Convert Date Columns

We convert the order and shipping date columns into datetime format.

Invalid date values are converted into missing values.

```python
new_batch_clean['Order_Date'] = pd.to_datetime(
    new_batch_clean['Order_Date'], errors='coerce'
)

new_batch_clean['Ship_Date'] = pd.to_datetime(
    new_batch_clean['Ship_Date'], errors='coerce'
)

print(new_batch_clean[['Order_Date', 'Ship_Date']].dtypes)
```

### 8.7 Validate Shipping Dates

We identify shipping dates that occur before their corresponding
order dates and replace these invalid dates with missing values.

```python
invalid_shipping = (
    new_batch_clean['Ship_Date'].notna() &
    new_batch_clean['Order_Date'].notna() &
    (new_batch_clean['Ship_Date'] < new_batch_clean['Order_Date'])
)

print("Invalid Shipping Dates:", invalid_shipping.sum())

new_batch_clean.loc[invalid_shipping, 'Ship_Date'] = pd.NaT
```

### 8.8 Final Check

We review the cleaned dataset to verify the results
and check the remaining missing values and data types.

```python
print("Cleaned New Batch Shape:", new_batch_clean.shape)

print("\nMissing Values:")
print(new_batch_clean.isnull().sum())

print("\nExact Duplicates:", new_batch_clean.duplicated().sum())

print("\nData Types:")
print(new_batch_clean.dtypes)

display(new_batch_clean.head())
```

## 9. Data Cleaning - Customers and Products

In this section, we clean the customers and products master datasets.

We will standardize text values, validate unique identifiers,
and check numerical columns to ensure data consistency.

```python
customers_clean = customers.copy()
products_clean = products.copy()

print("Customers Shape:", customers_clean.shape)
print("Products Shape:", products_clean.shape)
```

### 9.2 Clean Customers Dataset

We standardize the Region column by removing extra spaces
and applying consistent capitalization.

We also verify that Customer_ID contains unique identifiers.

```python
# Standardize Region
customers_clean['Region'] = (
    customers_clean['Region']
    .str.strip()
    .str.title()
)

# Check Customer_ID duplicates
print("Duplicate Customer IDs:",
      customers_clean['Customer_ID'].duplicated().sum())

print("\nRegion Values:")
print(customers_clean['Region'].value_counts())

display(customers_clean.head())
```

### 9.3 Validate Customers Dataset

We check the cleaned customer dataset for missing values,
duplicate rows, and duplicate customer identifiers.

```python
print("Missing Values:")
print(customers_clean.isnull().sum())

print("\nExact Duplicates:",
      customers_clean.duplicated().sum())

print("\nDuplicate Customer IDs:",
      customers_clean['Customer_ID'].duplicated().sum())
```

### 9.4 Clean Products Dataset

We standardize the Category column by removing extra spaces
and applying consistent capitalization.

We also validate product identifiers and numerical inventory values.

```python
# Standardize Category
products_clean['Category'] = (
    products_clean['Category']
    .str.strip()
    .str.title()
)

# Check Product_ID duplicates
print("Duplicate Product IDs:",
      products_clean['Product_ID'].duplicated().sum())

print("\nProduct Categories:")
print(products_clean['Category'].value_counts())

display(products_clean.head())
```

### 9.5 Validate Numerical Columns

We ensure that Unit_Cost, Current_Stock, and Reorder_Level
contain valid non-negative numerical values.

### 9.6 Final Validation

We perform a final validation of both master datasets
to ensure that the cleaning operations were successful.

```python
for name, df in {
    'Customers': customers_clean,
    'Products': products_clean
}.items():

    print(f"\n{'='*40}")
    print(name)
    print('='*40)

    print("Shape:", df.shape)
    print("Missing Values:", df.isnull().sum().sum())
    print("Exact Duplicates:", df.duplicated().sum())

display(customers_clean.head())
display(products_clean.head())
```

## 10. Data Cleaning - Previous Days Summary

We validate date formats, numerical columns, missing values,

## 11. Data Integration and Validation

In this section, we prepare the cleaned orders datasets for integration.

We standardize column names, data types, and identifier formats
before combining the historical orders with the new daily batch.

```python
# Make sure both datasets have the same columns in the same order
print("Same Columns:", list(orders_clean.columns) == list(new_batch_clean.columns))

# Standardize identifier columns
id_columns = ['Order_ID', 'Customer_ID', 'Product_ID']

for df in [orders_clean, new_batch_clean]:
    for col in id_columns:
        df[col] = df[col].astype('string').str.strip()

# Standardize numerical columns
numeric_columns = [
    'Quantity',
    'Unit_Price',
    'Discount',
    'Shipping_Cost'
]

for df in [orders_clean, new_batch_clean]:
    for col in numeric_columns:
        df[col] = pd.to_numeric(df[col], errors='coerce')

# Ensure both datasets use the same column order
new_batch_clean = new_batch_clean[orders_clean.columns]

print("Orders:", orders_clean.shape)
print("New Batch:", new_batch_clean.shape)
```

### 11.2 Combine Orders and New Daily Batch

We combine the cleaned historical orders and new daily batch
into one unified dataset using concatenation.

This creates a single dataset that can be used for further analysis.

```python
all_orders = pd.concat(
    [orders_clean, new_batch_clean],
    ignore_index=True
)

print("Combined Dataset Shape:", all_orders.shape)

display(all_orders.head())
```

### 11.3 Validate Customer IDs

We compare the Customer_ID values in the combined orders dataset
with the customer identifiers in the customers master dataset.

This helps identify orders that contain missing or unknown customer IDs.

```python
valid_customer_ids = customers_clean['Customer_ID'].dropna().unique()

missing_customer_id = all_orders['Customer_ID'].isna()

unknown_customer_id = (
    all_orders['Customer_ID'].notna() &
    ~all_orders['Customer_ID'].isin(valid_customer_ids)
)

print("Missing Customer IDs:", missing_customer_id.sum())
print("Unknown Customer IDs:", unknown_customer_id.sum())

print("\nUnknown Customer Values:")
print(all_orders.loc[unknown_customer_id, 'Customer_ID'].unique())
```

### 11.4 Validate Product IDs

We compare the Product_ID values in the combined orders dataset
with the identifiers available in the products master dataset.

This helps identify missing or invalid product references.

```python
valid_product_ids = products_clean['Product_ID'].dropna().unique()

missing_product_id = all_orders['Product_ID'].isna()

unknown_product_id = (
    all_orders['Product_ID'].notna() &
    ~all_orders['Product_ID'].isin(valid_product_ids)
)

print("Missing Product IDs:", missing_product_id.sum())
print("Unknown Product IDs:", unknown_product_id.sum())

print("\nUnknown Product Values:")
print(all_orders.loc[unknown_product_id, 'Product_ID'].unique())
```

```python
customers_clean['Customer_ID'] = (
    customers_clean['Customer_ID'].astype('string').str.strip()
)

products_clean['Product_ID'] = (
    products_clean['Product_ID'].astype('string').str.strip()
)
```

### 11.5 Merge Orders with Master Data

We enrich the combined orders dataset by joining it with
the customers and products master datasets.

We use left joins to preserve all order records,
including those with missing or unmatched identifiers.

```python
orders_enriched = all_orders.merge(
    customers_clean,
    on='Customer_ID',
    how='left'
)

orders_enriched = orders_enriched.merge(
    products_clean,
    on='Product_ID',
    how='left'
)

print("Enriched Dataset Shape:", orders_enriched.shape)

display(orders_enriched.head())
```

## Final Data Quality Validation

Before exporting the cleaned datasets, we perform a final quality check.

We verify duplicate identifiers, unmatched customer and product IDs,
missing values, and dataset dimensions to ensure data consistency.

```python
# Check duplicate IDs in master datasets
print("Duplicate Customer IDs:",
      customers_clean['Customer_ID'].duplicated().sum())

print("Duplicate Product IDs:",
      products_clean['Product_ID'].duplicated().sum())

# Check unmatched IDs
print("\nMissing Customer IDs:", all_orders['Customer_ID'].isna().sum())
print("Missing Product IDs:", all_orders['Product_ID'].isna().sum())

print("Unknown Customer IDs:",
      (~all_orders['Customer_ID'].isin(
          customers_clean['Customer_ID']
      ) & all_orders['Customer_ID'].notna()).sum())

print("Unknown Product IDs:",
      (~all_orders['Product_ID'].isin(
          products_clean['Product_ID']
      ) & all_orders['Product_ID'].notna()).sum())

# Check row counts
print("\nCombined Orders:", len(all_orders))
print("Enriched Orders:", len(orders_enriched))

# Check duplicates
print("Duplicate Rows:", all_orders.duplicated().sum())
```

```python
print("Orders Before Merge:", len(all_orders))
print("Orders After Merge:", len(orders_enriched))

if len(all_orders) == len(orders_enriched):
    print("Merge Validation Passed")
else:
    print("Warning: Merge Changed the Number of Rows")
```

```python
# Orders with missing Customer_ID
missing_customers = all_orders[
    all_orders['Customer_ID'].isna()
]

# Orders with unknown Customer_ID
unknown_customers = all_orders[
    all_orders['Customer_ID'].notna() &
    ~all_orders['Customer_ID'].isin(
        customers_clean['Customer_ID']
    )
]

# Orders with unknown Product_ID
unknown_products = all_orders[
    all_orders['Product_ID'].notna() &
    ~all_orders['Product_ID'].isin(
        products_clean['Product_ID']
    )
]

print("Missing Customer Records:")
display(missing_customers)

print("Unknown Customer Records:")
display(unknown_customers)

print("Unknown Product Records:")
display(unknown_products)
```

```python
print("Unknown Customer IDs:")
print(unknown_customers['Customer_ID'].unique())

print("\nUnknown Product IDs:")
print(unknown_products['Product_ID'].unique())
```


### 11.7 Identify and Remove Invalid ID Records

In this step, we identify orders with missing or unmatched Customer IDs and Product IDs.

Records with invalid references will be excluded from the final analysis dataset, while the original cleaned datasets will remain unchanged.


```python

# Keep only orders with valid Customer IDs and Product IDs

valid_orders = orders_enriched[
    (orders_enriched['Customer_ID_Status'] == 'Valid') &
    (orders_enriched['Product_ID_Status'] == 'Valid')
].copy()

print("Rows before removal:", len(orders_enriched))
print("Rows after removal:", len(valid_orders))
print("Removed rows:", len(orders_enriched) - len(valid_orders))
```


### 11.7 Identify Invalid Records

In this step, we identify records with missing or unmatched Customer IDs and Product IDs to prepare a valid dataset for analysis.


```python

# Identify Customer ID status

orders_enriched['Customer_ID_Status'] = np.where(
    orders_enriched['Customer_ID'].isna(),
    'Missing',
    np.where(
        orders_enriched['Customer_ID'].isin(customers_clean['Customer_ID']),
        'Valid',
        'Unmatched'
    )
)

# Identify Product ID status

orders_enriched['Product_ID_Status'] = np.where(
    orders_enriched['Product_ID'].isna(),
    'Missing',
    np.where(
        orders_enriched['Product_ID'].isin(products_clean['Product_ID']),
        'Valid',
        'Unmatched'
    )
)

print(orders_enriched[
    ['Customer_ID_Status', 'Product_ID_Status']
].value_counts())
```


### 11.8 Create the Final Valid Dataset

Records with missing or unmatched Customer IDs or Product IDs will be excluded from the final analysis dataset. The original datasets will remain unchanged.


```python

# Keep only records with valid Customer and Product IDs

valid_orders = orders_enriched[
    (orders_enriched['Customer_ID_Status'] == 'Valid') &
    (orders_enriched['Product_ID_Status'] == 'Valid')
].copy()

print("Original Records:", len(orders_enriched))
print("Final Valid Records:", len(valid_orders))
print("Removed Records:", len(orders_enriched) - len(valid_orders))
```

```python
print("Combined Orders Shape:", all_orders.shape)
print("Enriched Orders Shape:", orders_enriched.shape)

print("\nExact Duplicates in Combined Orders:")
print(all_orders.duplicated().sum())

print("\nMissing Values in Combined Orders:")
print(all_orders.isnull().sum())

print("\nMissing Values in Enriched Orders:")
print(orders_enriched.isnull().sum())
```


### 11.9 Final Data Validation

In this step, we perform a final validation of the dataset to ensure that the final analysis data contains no missing or unmatched Customer IDs or Product IDs, and no duplicate records.


```python

# Final validation checks

print("Final Dataset Shape:", valid_orders.shape)

print("\nMissing Customer IDs:",
      valid_orders['Customer_ID'].isna().sum())

print("Missing Product IDs:",
      valid_orders['Product_ID'].isna().sum())

print("Invalid Customer IDs:",
      (valid_orders['Customer_ID_Status'] != 'Valid').sum())

print("Invalid Product IDs:",
      (valid_orders['Product_ID_Status'] != 'Valid').sum())

print("Duplicate Rows:", valid_orders.duplicated().sum())
```


## 12. Export Cleaned Datasets

After completing the cleaning, integration, and validation steps, we export all cleaned datasets into CSV files.

The final valid orders dataset will also be exported separately for further analysis.


```python

import os
import shutil

# Create output folder
os.makedirs('cleaned_data', exist_ok=True)

# Export cleaned datasets
orders_clean.to_csv('cleaned_data/cleaned_orders.csv', index=False)
new_batch_clean.to_csv('cleaned_data/cleaned_new_daily_batch.csv', index=False)
customers_clean.to_csv('cleaned_data/cleaned_customers.csv', index=False)
products_clean.to_csv('cleaned_data/cleaned_products.csv', index=False)
summary_clean.to_csv('cleaned_data/cleaned_previous_days_summary.csv', index=False)

# Export combined and enriched datasets
all_orders.to_csv('cleaned_data/combined_orders.csv', index=False)
orders_enriched.to_csv('cleaned_data/enriched_orders.csv', index=False)

# Export final valid dataset
valid_orders.to_csv('cleaned_data/final_valid_orders.csv', index=False)

# Create final quality report
final_report = pd.DataFrame({
    'Metric': [
        'Original Orders',
        'Original New Batch',
        'Cleaned Orders',
        'Cleaned New Batch',
        'Combined Orders',
        'Enriched Orders',
        'Final Valid Orders',
        'Removed Invalid Records',
        'Missing Customer IDs in Final Dataset',
        'Missing Product IDs in Final Dataset',
        'Duplicate Rows in Final Dataset'
    ],
    'Value': [
        len(orders),
        len(new_batch),
        len(orders_clean),
        len(new_batch_clean),
        len(all_orders),
        len(orders_enriched),
        len(valid_orders),
        len(orders_enriched) - len(valid_orders),
        valid_orders['Customer_ID'].isna().sum(),
        valid_orders['Product_ID'].isna().sum(),
        valid_orders.duplicated().sum()
    ]
})

final_report.to_csv('cleaned_data/final_quality_report.csv', index=False)

# Create ZIP file
if os.path.exists('cleaned_data_final.zip'):
    os.remove('cleaned_data_final.zip')

shutil.make_archive('cleaned_data_final', 'zip', 'cleaned_data')

print("Export completed successfully!")
print("Number of exported files:", len(os.listdir('cleaned_data')))
print("Final valid orders:", len(valid_orders))
```
